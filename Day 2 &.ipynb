### Coloumn Filtering

```python
df.select("Customer_City").distinct().show()
df.select("Product_category").distinct().show()
df.select("Agent_name").distinct().show()
```

```python
selected_df = df.select("Customer_City", "Product_category", "Agent_name")
filtered_df = selected_df.filter(selected_df.Product_category == "Electronics")
```

```python
# Select specific columns
selected_df = df.select("Customer_City", "Product_category", "Agent_name", )

# Filter rows where Product_category is 'Electronics'
filtered_df = selected_df.filter(selected_df.Product_category == "Electronics")

# Group by Customer_City and count issues
grouped_df = filtered_df.groupBy("Customer_City").count()

# Order by count descending
ordered_df = grouped_df.orderBy("count", ascending=False)

display(ordered_df)
```
